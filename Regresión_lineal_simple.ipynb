```python
import numpy as np

import matplotlib.pyplot as plt

np.random.seed(76)
n= 100
x= np.random.randint(0, 100, size=n)
E= np.random.normal(loc=0, scale=10, size=n)
y =2 *x + 3 + E

x_= np.mean(x)
y_ = np.mean(y)
cov_xy = np.sum((x - x_) * (y - y_))
var_x_sum = np.sum((x - x_) ** 2)


beta_1 = cov_xy / var_x_sum
beta_0 = y_ - beta_1 * x_

y_pred = beta_0 + beta_1 * x

sst = np.sum((y - y_) ** 2)
sse = np.sum((y - y_pred) ** 2)
r2 = 1 - (sse / sst)
mse = sse / n


var_x = np.var(x, ddof=1)

print(f"Varianza de x: {var_x:.4f}")
print(f"Beta 0 (intersección): {beta_0:.4f}")
print(f"Beta 1 (pendiente): {beta_1:.4f}")
print(f"SST: {sst:.4f}")
print(f"SSE: {sse:.4f}")
print(f"R^2: {r2:.4f}")
print(f"MSE: {mse:.4f}")

plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6, label="Datos")
plt.plot(
    x, y_pred, color="red", linewidth=2, label=f"Línea MCO: y = {beta_1:.2f}x + {beta_0:.2f}"
)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Regresión Lineal Simple")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
